## Download NAIP Imagery and Building footprint

Fetch NAIP aerial imagery for the specified bounding box. The `max_items` parameter limits the number of downloaded files.

```python
import json
from shapely.geometry import Polygon, mapping

for state_code, bbox in state_bboxes.items():
    print(f"\n=== Processing {state_code} ===")

    # -------------------------------
    # 1. Download NAIP
    # -------------------------------
    naip_files = download_naip(
        bbox=bbox,
        output_dir=f"data/{state_code}/naip",
        max_items=1
    )
    print(f"✓ NAIP downloaded for {state_code}")

    # -------------------------------
    # 2. Download Buildings
    # -------------------------------
    buildings_path = f"data/{state_code}/buildings.geojson"

    try:
        buildings = download_overture_buildings(
            bbox=bbox,
            output=buildings_path
        )
        print(f"✓ Buildings downloaded for {state_code}: {buildings_path}")

    except FileNotFoundError:
        print(f"⚠ No Overture buildings found for {state_code}. Creating empty file...")

        # Write an EMPTY GeoJSON FeatureCollection
        empty = {"type": "FeatureCollection", "features": []}

        os.makedirs(os.path.dirname(buildings_path), exist_ok=True)
        with open(buildings_path, "w") as f:
            json.dump(empty, f)

        print(f"✓ Empty buildings.geojson created for {state_code}")

print("\n🚀 Completed downloading NAIP + buildings for all 50 states!")
```
